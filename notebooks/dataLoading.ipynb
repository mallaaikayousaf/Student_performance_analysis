```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/StudentsPerformance.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
categorical = ['gender', 'race/ethnicity', 'parental level of education', 'lunch', 'test preparation course']
for col in categorical:
    print(f"\n{col} unique values:")
    print(df[col].unique())
```

```python
subjects = ['math score', 'reading score', 'writing score']
```

```python
print("\n=== MEAN SCORES ===")
for subject in subjects:
    print(f"{subject}: {df[subject].mean():.2f}")
```

```python
print("\n=== MEDIAN SCORES ===")
for subject in subjects:
    print(f"{subject}: {df[subject].median():.2f}")
```

```python
print("\n=== STANDARD DEVIATION ===")
for subject in subjects:
    print(f"{subject}: {df[subject].std():.2f}")
```

```python
print("\n=== MIN AND MAX SCORES ===")
for subject in subjects:
    print(f"{subject} - Min: {df[subject].min()}, Max: {df[subject].max()}")
```
