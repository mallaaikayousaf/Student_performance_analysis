```python
import pandas as pd
import numpy as np

df = pd.read_csv("../Data/StudentsPerformance.csv")

print("="*60)
print("DESCRIPTIVE STATISTICS ANALYSIS")
print("="*60)
```

```python
subjects = ['math score', 'reading score', 'writing score']

results = []

for subject in subjects:
    mean_val = df[subject].mean()
    median_val = df[subject].median()
    mode_val = df[subject].mode().iloc[0]
    std_val = df[subject].std()
    min_val = df[subject].min()
    max_val = df[subject].max()
    
    results.append({
        'Subject': subject,
        'Mean': mean_val,
        'Median': median_val,
        'Mode': mode_val,
        'Std Dev': std_val,
        'Min': min_val,
        'Max': max_val
    })

stats_df = pd.DataFrame(results)
print("\n" + "="*60)
print("FULL DESCRIPTIVE STATISTICS")
print("="*60)
print(stats_df.round(2))
```

```python
print("\n" + "="*60)
print("MEAN VS MEDIAN COMPARISON")
print("="*60)

for subject in subjects:
    mean_val = df[subject].mean()
    median_val = df[subject].median()
    diff = mean_val - median_val
    
    print(f"\n{subject}:")
    print(f"  Mean: {mean_val:.2f}")
    print(f"  Median: {median_val:.2f}")
    print(f"  Difference: {diff:.2f}")
    
    if diff > 5:
        print(f"  → Right skewed (higher mean suggests exceptional high scores)")
    elif diff < -5:
        print(f"  → Left skewed (higher median suggests consistently good scores)")
    else:
        print(f"  → Approximately symmetric distribution")
```

```python
print("\n" + "="*60)
print("ANALYSIS BY GENDER")
print("="*60)

gender_stats = df.groupby('gender')[subjects].mean()
print(gender_stats.round(2))

print("\nDetailed statistics by gender:")
gender_detailed = df.groupby('gender')[subjects].agg(['mean', 'median', 'std', 'count'])
print(gender_detailed.round(2))
```

```python
print("\n" + "="*60)
print("ANALYSIS BY PARENTAL LEVEL OF EDUCATION")
print("="*60)

parent_stats = df.groupby('parental level of education')[subjects].mean()
print(parent_stats.round(2))

# Sort by math score
parent_sorted = parent_stats.sort_values('math score', ascending=False)
print("\nSorted by Math Score (highest to lowest):")
print(parent_sorted.round(2))
```

```python
print("\n" + "="*60)
print("ADDITIONAL GROUPBY ANALYSES")
print("="*60)

# By test preparation
print("\nBy Test Preparation Course:")
test_prep = df.groupby('test preparation course')[subjects].mean()
print(test_prep.round(2))

# By lunch type
print("\nBy Lunch Type:")
lunch_stats = df.groupby('lunch')[subjects].mean()
print(lunch_stats.round(2))

# By race/ethnicity
print("\nBy Race/Ethnicity:")
race_stats = df.groupby('race/ethnicity')[subjects].mean()
print(race_stats.round(2))
```
